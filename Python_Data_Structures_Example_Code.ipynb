```
# A complete Python script demonstrating lists, arrays, and dictionaries

print("--- Demonstrating Lists ---")
# 1. Creating a list
fruits = ["apple", "banana", "cherry"]
print(f"Initial list: {fruits}")

# 2. Accessing elements
print(f"First element: {fruits[0]}")
print(f"Sliced elements: {fruits[1:3]}")

# 3. Modifying an element
fruits[1] = "blueberry"
print(f"List after modifying: {fruits}")

# 4. Adding elements
fruits.append("orange")
fruits.insert(0, "grape")
print(f"List after adding: {fruits}")

# 5. Removing elements
fruits.remove("cherry")
popped_item = fruits.pop(2)
print(f"List after removing 'cherry' and popping '{popped_item}': {fruits}")

print("\n" + "="*30 + "\n")

# ---
# 6. Demonstrating Arrays
# For this, we need the built-in 'array' module.
import array as arr

print("--- Demonstrating Arrays ---")
# Create an array of signed integers ('i')
try:
    # This will work
    my_array = arr.array('i', [10, 20, 30])
    print(f"Initial array: {my_array}")

    # Adding a new integer element
    my_array.append(40)
    print(f"Array after appending: {my_array}")

    # This would cause an error because arrays are type-specific
    # my_array.append("hello") # Uncommenting this would raise an error
except TypeError as e:
    print(f"Error caught: {e}")

print("\n" + "="*30 + "\n")

# ---
# NEW ADDITION: Demonstrating why arrays don't work with strings
print("--- Array with String Data (Incorrect) ---")
try:
    # This will cause a TypeError because 'u' is for unicode characters, not strings
    string_array_attempt = arr.array('u', ['h', 'e', 'l', 'l', 'o'])
    print(f"Attempted array of strings: {string_array_attempt}")
except TypeError as e:
    print(f"Error caught: {e}")
    print("This error occurs because the 'array' module is for primitive numeric types, not strings.")

print("\n" + "="*30 + "\n")

# ---
# Demonstrating the correct way to handle collections of strings
print("--- Correctly Handling Strings with a List ---")
string_list = ["apple", "banana", "cherry", "grape"]
print(f"A list of strings: {string_list}")
string_list.append("orange")
print(f"List after appending a new string: {string_list}")

print("\n" + "="*30 + "\n")

# ---
# 7. Demonstrating Dictionaries
print("--- Demonstrating Dictionaries ---")
# Creating a dictionary
person = {
    "name": "Bob",
    "age": 25,
    "is_student": True
}
print(f"Initial dictionary: {person}")

# 8. Accessing values
print(f"Person's name: {person['name']}")

# 9. Adding and modifying key-value pairs
person["age"] = 26
person["city"] = "San Francisco"
print(f"Dictionary after updating and adding: {person}")

# 10. Removing a key-value pair
del person["is_student"]
print(f"Dictionary after deleting 'is_student': {person}")

# 11. Looping through a dictionary
print("Iterating through dictionary items:")
for key, value in person.items():
    print(f"  {key}: {value}")
```

```
# Demonstrating a dictionary as a 'struct'
print("--- Demonstrating a Dictionary as a 'Struct' ---")

# Create a dictionary to represent a student
student = {
    "student_id": "S12345",
    "name": "Alice Smith",
    "major": "Computer Science",
    "gpa": 3.8,
    "courses": ["CS101", "MATH203", "PHYS101"]
}

print(f"Student information: {student}")

# Accessing individual fields
print(f"Student ID: {student['student_id']}")
print(f"Student Name: {student['name']}")
print(f"Student Major: {student['major']}")
print(f"Student GPA: {student['gpa']}")
print(f"Student Courses: {student['courses']}")

# Adding a new field
student["year"] = 3
print(f"Student information after adding year: {student}")

# Modifying a field
student["gpa"] = 3.9
print(f"Student information after updating GPA: {student}")

# Accessing elements within a list field
print(f"First course: {student['courses'][0]}")
```
